## Problema da Mochila utilizando Heurística Gulosa

### Introdução

O Problema da Mochila consiste em selecionar itens para uma mochila de capacidade limitada, buscando maximizar o valor total transportado sem ultrapassar o limite de peso.

Neste trabalho foi utilizada uma heurística gulosa baseada na relação valor/peso dos itens.

```python
itens = [
    {"nome": "Notebook", "peso": 3, "valor": 2000},
    {"nome": "Celular", "peso": 1, "valor": 1500},
    {"nome": "Livro", "peso": 2, "valor": 200},
    {"nome": "Fone", "peso": 1, "valor": 300},
    {"nome": "Tablet", "peso": 2, "valor": 1200}
]

capacidade = 5

print("Dados carregados!")
```

```python
for item in itens:
    item["razao"] = item["valor"] / item["peso"]

for item in itens:
    print(item)
```

```python
itens.sort(
    key=lambda item: item["razao"],
    reverse=True
)

for item in itens:
    print(item["nome"], item["razao"])
```

```python
peso_total = 0
valor_total = 0
selecionados = []

for item in itens:

    if peso_total + item["peso"] <= capacidade:

        selecionados.append(item)

        peso_total += item["peso"]

        valor_total += item["valor"]
```

```python
print("Itens escolhidos:\n")

for item in selecionados:
    print(
        f'{item["nome"]} - Peso: {item["peso"]} - Valor: R${item["valor"]}'
    )

print("\nResumo")
print(f"Peso total: {peso_total}")
print(f"Valor total: R${valor_total}")
```

## Conclusão

A heurística gulosa foi capaz de encontrar uma solução viável para o Problema da Mochila.

O método selecionou os itens de melhor relação valor/peso e respeitou a capacidade máxima da mochila.

Apesar de não garantir a solução ótima em todos os casos, apresenta baixo custo computacional e fácil implementação.
